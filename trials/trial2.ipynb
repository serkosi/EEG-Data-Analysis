```python
import pyedflib
import numpy as np
import os
```

```python
root = ("E:\\PhD\\materials_neuroscience\\trials\\VAL_KLAUS")
list_file_names = []
for path, dirs, files in os.walk(root):
    for file in files:
        if file.endswith(".bdf"):
             list_file_names.append(os.path.join(root, file))
len(list_file_names)
```

```python
file_name = pyedflib.data.get_generator_filename()
list_sigbufs = np.empty((len(list_file_names), 0)).tolist()
for j in range(len(list_file_names)):
    f = pyedflib.EdfReader(list_file_names[j])
    n = f.signals_in_file
    signal_labels = f.getSignalLabels()
    list_sigbufs[j] = np.zeros((n, f.getNSamples()[0]))
    for i in np.arange(n):
            (list_sigbufs[j])[i, :] = f.readSignal(i)
```
